```python
import pandas as pd

import numpy as np

import matplotlib.pyplot as plt

df=pd.read_csv("data/Titanic-Dataset.csv")

df.head()   
```

```python
df.tail()
```

```python
ages=df['Age'].to_numpy()
```

```python
print(ages)
```

```python
clean=ages[~np.isnan(ages)]
print(clean)
```

```python
Ages=clean
```

```python
print(Ages)
```

```python
x=Ages[0:10]
print(x)
```

```python
a=df['Fare'].to_numpy()
print(a)
```

```python
Fares= a[~np.isnan(a)]
print(Fares)
```

```python
y=Fares[0:10]
print(y)
```

##  Plots

```python
plt.scatter(x,y)
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.show()
```

```python
plt.scatter(x,y,color='orange')
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.show()
```

```python


plt.plot(x,y)
plt.grid(True)
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.show()
```

```python


plt.plot(x,y,marker='o')
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.show()
```

```python

plt.plot(x,y,marker='o')
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.gca().set_facecolor("lightyellow")
plt.show()
```
